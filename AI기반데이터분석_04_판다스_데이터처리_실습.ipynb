## 1. 파이썬 파일 처리

```python
sample_file = 'data/sample.txt'
```

```python
# 파일을 open하여 전체 내용을 읽어서 출력
with open(sample_file, 'rt') as f:
    content = f.read()
    print(content)
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기

with open(sample_file, 'r') as f:
    lines = f.readlines()
    for line in lines:
        print(line, end="")
        
print("----- 끝 -----")
```

```python
with open(sample_file, 'r') as f:
    line = f.readline()
    while line:
        print(line, end="")
        line = f.readline()
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옵
with open(sample_file, 'r') as f:
    for line in f:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 여러 개의 파일 읽기 (모든 txt 파일을 읽어서 출력하기)
import glob

txt_file_pattern = "data/*.txt"
files = glob.glob(txt_file_pattern)
print(files)
for file in files:
    with open(file, 'r') as f:
        content = f.read()
        print(content)

print("----- 끝 -----")
```

### 4-4. 데이터 접근/추출
* 범위 지정 : 직접 지정, 인덱서 이용
* 조건 지정

```python
score_dict={'이름': ['정약용', '이순신', '이율곡', '홍길동'],
            '국어' : [70, 100, 90, 85],
            '영어' : [80, 80, 95, 100],
            '수학' : [100, 70, 90, 90]}
score_df = pd.DataFrame(score_dict)
score_df
```

```python
# 행 지정 -> 슬라이싱
score_df[0:2]
```

```python
# 열 접근
score_df['이름']
```

```python
# 특정 (행,열)의 데이터 접근
score_df.loc[0, '국어']
```

```python
# 인덱서 이용 : 레이블(이름) 인덱서
score_df = score_df.set_index('이름')
```

```python
# 행 접근
score_df.loc['이순신']
```

```python
# 열 접근
score_df.loc[:, '국어']
```

```python
# (행,열) 접근
score_df.loc['이순신', '국어']
```

```python
# 인덱서 이용 : 위치(숫자) 인덱서
# 행 접근
score_df.iloc[0]
```

```python
# 열 접근
score_df.iloc[:, 1]
```

```python
# (행,열) 접근
score_df.iloc[0, 1]
```

```python
# 조건 지정 추출 : Boolean Indexing
# 직접 지정 방식으로 행 추출
score_df[score_df['국어'] >= 90]
```

```python
# 인덱서 방식으로 행 추출
score_df.loc[score_df['국어'] >= 90]
```

```python
# (행,열) 데이터 추출
score_df.loc[score_df['국어'] >= 90, ['영어', '수학']]
```

```python
# 열 데이터에 조건 지정
score_df[(score_df['국어'] >= 90) & (score_df['수학'] >= 90)]
```

## 4-5. 그룹별로 요약하여 데이터 확인
* value_counts()
* groupby()
* pivot_table()

```python
sample_file = 'data/sales_sample.csv'
```

```python
# 샘플 파일 로딩
sales_df = pd.read_csv('sales_sample.csv')
```

```python
# 지역별 판매 기록의 개수
sales_df['지역'].value_counts()
```

```python
# 상품별 판매 기록의 개수와 비율
sales_df['상품'].value_counts(normalize=True)
```

```python
# 지역과 상품의 조합별 판매 기록 개수
sales_df.groupby(['지역', '상품']).size()
```

```python
# 지역별 총매출
sales_df.groupby('지역')['매출'].sum()
```

```python
# 상품별 평균 매출
sales_df.groupby('상품')['매출'].mean()
```

```python
# 지역별 판매수량의 합계와 매출의 합계
sales_df.groupby('지역')[['판매수량', '매출']].sum()
```

```python
# 지역과 상품별 평균 매출
sales_df.groupby(['지역', '상품'])['매출'].mean()
```

```python
# 지역별·상품별 총매출을 구하고 행과 열의 합계 표시
sales_df.pivot_table(index='지역', columns='상품', values='매출', aggfunc='sum', margins=True)
```
